```python
# Import required libraries
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Load MNIST dataset
data = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=transforms.ToTensor()
)

# Create DataLoader
loader = DataLoader(
    data,
    batch_size=64,
    shuffle=True
)

# Autoencoder Model
class Autoencoder(nn.Module):

    def __init__(self):
        super(Autoencoder, self).__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 128),
            nn.ReLU(),
            nn.Linear(128, 32)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(32, 128),
            nn.ReLU(),
            nn.Linear(128, 784),
            nn.Sigmoid()
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded


# Initialize model
model = Autoencoder()

# Loss function
loss_fn = nn.MSELoss()

# Optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

# Training
for epoch in range(10):

    total_loss = 0

    for images, _ in loader:

        # Forward pass
        output = model(images)

        # Flatten original images
        images = images.view(images.size(0), -1)

        # Calculate loss
        loss = loss_fn(output, images)

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(
        f"Epoch {epoch+1}/10, "
        f"Loss: {total_loss/len(loader):.4f}"
    )

print("\nAutoencoder training completed!")
```
